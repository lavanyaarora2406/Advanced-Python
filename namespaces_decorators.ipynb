### Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces:
- Builtin Namespace
- Global Namespace
- Enclosing Namespace
- Local Namespace

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

LEGB rule = local enclosing global builtin 

```python
# local and global
# global var
a = 2 # global scope

def temp():
  # local var
  b = 3 # local scope 
  print(b)

temp()
print(a)
```

```python
# local and global -> same name
a = 2 # global scope 

def temp():
  # local var
  a = 3 # local scope 
  print(a)

temp()
print(a)
```

```python
# local and global -> local does not have but global has
a = 2

def temp():
  # local var
  print(a)

temp()
print(a)
```

```python
# local and global -> editing global
a = 2

def temp():
  # local var
  a += 1 # we cannot edit global variable inside local scope without declaring it as global
  print(a)

temp()
print(a)
```

```python
a = 2

def temp():
  # local var
  global a # changing the global a ... 
  a += 1
  print(a)

temp()
print(a)
```

```python
# local and global -> global created inside local
def temp():
  # local var
  global a
  a = 1
  print(a)

temp()
print(a)
```

```python
# local and global -> function parameter is local
def temp(z):
  # local var
  print(z)

a = 5
temp(5)
print(a)
print(z)    
```

```python
# Enclosing scope = it happens in nested functions ....
def outer():
  def inner():
    print('inner function')
  inner()
  print('outer function')

# outer one  scope is enclosing scope for inner function

outer()
print('main program')
```

```python
# nonlocal keyword
def outer(): # this is local variable for outer function but enclosing scope for inner function
  
  a = 1
  def inner():
    nonlocal a # this is local variable for outer function but enclosing scope for inner function
    a += 1
    print('inner',a)
  inner()
  print('outer',a)


outer()
print('main program')
# we make a nested fucntion so outer one would be enclosing scope for inner function and we can use nonlocal keyword to access the variable of outer function inside inner function
```

### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python
- `Built in decorators` like `@staticmethod`, `@classmethod`, `@abstractmethod` and `@property` etc
- `User defined decorators` that we programmers can create according to our needs

```python
# Python are 1st class function

def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

```python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello neha')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()

# closure = function object that remembers values in enclosing scopes even if they are not present in memory
# means it remmbers the parent function's variable even after the parent function has finished execution due to return statement 
# on this concept of closure the decorators is Based ..

```

```python
# Better syntax?
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

@my_decorator # decorator syntax  
def hello():
  print('hello')

hello()
```

```python

# decorator that tell the execution time of the function it decorates
import time


def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print('time taken by',func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello wolrd')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

@timer
def power(a,b):
  print(a**b)

hello()
square(2)
power(2,3)
```
